```python
import nibabel as nib

image_filename = "train/P1/T1/P1_T1_FLAIR.nii.gz"
img = nib.load(image_filename)
Shape = img.shape
data = img.get_fdata()

image_filename = "train/P1/T1/P1_T1_MASK.nii.gz"
img = nib.load(image_filename)
Mask = img.get_fdata()

image_filename = "train/P1/T1/P1_T1_T1.nii.gz"
img = nib.load(image_filename)
t1t1 = img.get_fdata()

image_filename = "train/P1/T1/P1_T1_T2.nii.gz"
img = nib.load(image_filename)
t1t2 = img.get_fdata()

```
